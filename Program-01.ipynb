```python
#code from chatgpt
# Import libraries
import numpy as np
import pandas as pd

# -------------------------
# 1. BASIC NUMPY OPERATIONS
# -------------------------

# Create NumPy arrays
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("Array a:", a)
print("Array b:", b)

# Arithmetic operations
print("Addition:", a + b)
print("Subtraction:", b - a)
print("Multiplication:", a * b)
print("Division:", b / a)

# Other basic operations
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))

# Slicing NumPy array
print("First 3 elements:", a[:3])
print("Last 2 elements:", a[-2:])


# -------------------------
# 2. CREATE PANDAS DATAFRAME
# -------------------------

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)


# -------------------------
# 3. INSPECT DATAFRAME
# -------------------------

print("\nFirst 2 rows:")
print(df.head(2))

print("\nDataFrame information:")
print(df.info())

print("\nStatistical summary:")
print(df.describe())


# -------------------------
# 4. INDEXING
# -------------------------

# Select one column
print("\nNames:")
print(df["Name"])

# Select multiple columns
print("\nName and Marks:")
print(df[["Name", "Marks"]])

# Select row by position
print("\nFirst row:")
print(df.iloc[0])

# Select row by index
print("\nRow with index 2:")
print(df.loc[2])


# -------------------------
# 5. SLICING
# -------------------------

# First 2 rows
print("\nFirst 2 rows:")
print(df.iloc[:2])

# Rows 1 to 3
print("\nRows 1 to 3:")
print(df.iloc[1:4])

# Rows and selected columns
print("\nRows 1-3, Name and Marks:")
print(df.loc[1:3, ["Name", "Marks"]])


# -------------------------
# 6. FILTERING
# -------------------------

# Students with marks greater than 80
print("\nStudents with marks > 80:")
print(df[df["Marks"] > 80])
```

```python
#LAB Code
import numpy as np
import pandas as pd

# NumPy
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print(a + b)
print(a * b)
print(np.mean(a))
print(a[:3])       # slicing

# Pandas DataFrame
df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Age": [20, 21, 22],
    "Marks": [80, 90, 85]
})

print(df)           # display
print(df.head())    # inspect
print(df["Name"])   # column indexing
print(df.iloc[0])   # row indexing
print(df.iloc[:2])  # slicing
print(df[df["Marks"] > 80])  # filtering
```
